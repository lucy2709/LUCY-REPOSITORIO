Operación LÓGICA NOT

```python
import numpy as np

def activar(valor):
    return 1 if valor >= 0 else 0

# Casos de la operación NOT
entradas = np.array([[0], [1]])
salida_correcta = np.array([1, 0])  # NOT 0 = 1, NOT 1 = 0

peso = -1
umbral = 0.5

print("Operación NOT")
print("-" * 40)

for e, esperado in zip(entradas, salida_correcta):
    suma = e[0]*peso + umbral
    resultado = activar(suma)
    print(f"Entrada: {e[0]} → Resultado: {resultado} → Debe ser: {esperado}")
```

Esta neurona invierte lo que recibe. Si entra un 1, lo multiplica por un número negativo y la suma baja, entonces no se activa y da 0. Si entra un 0, el peso no afecta y el umbral hace que la suma sea suficiente, así que se activa y da 1. Es decir, hace lo contrario de lo que encuentra: si le llega un "sí", dice "no", y si le llega un "no", dice "sí". Eso es justo lo que hace la operación NOT.
