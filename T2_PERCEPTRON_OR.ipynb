Operación LÓGICA OR

```python
# Importamos la librería
import numpy as np

# Función de activación: devuelve 1 si la suma es mayor o igual a 0, sino 0
def activacion(suma):
    return 1 if suma >= 0 else 0

# Datos de entrada y salida esperada para la operación OR
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_esperado = np.array([0, 1, 1, 1])

# Pesos y sesgo elegidos para que cumplan la regla OR
pesos = np.array([1, 1])
sesgo = 0

print("=== OPERACIÓN OR con 1 neurona ===")
print("Entradas | Suma ponderada | Salida | Esperado")
print("-" * 55)

for entrada, salida_real in zip(X, y_esperado):
    suma = np.dot(entrada, pesos) + sesgo
    salida_obtenida = activacion(suma)
    print(f"  {entrada}   |      {suma}        |    {salida_obtenida}    |    {salida_real}")
```
