```python
import numpy as np

# 1. Datos de entrada — Compuerta lógica AND
X = np.array([
    [0, 0],   # Ambas entradas apagadas
    [0, 1],   # Solo la segunda encendida
    [1, 0],   # Solo la primera encendida
    [1, 1]    # Ambas encendidas
])

Y = np.array([0, 0, 0, 1])  # Salida correcta: SOLO 1 cuando ambas son 1

# 2. Pesos y umbral
W = np.array([1, 1])  # Cada entrada vale 1 punto
umbral = 2             # Necesitamos sumar 2 para activarse

# 3. Clase: Neurona de McCulloch-Pitts
class McCullochPitts:
    def __init__(self, weights, umbral):
        self.weights = weights   # Guardamos los pesos
        self.umbral = umbral     # Guardamos el umbral

    def predict(self, X):
        predictions = []
        # Recorremos cada par de entradas
        for i in range(X.shape[0]):
            # Suma ponderada = entrada1*peso1 + entrada2*peso2
            weighted_sum = np.dot(self.weights, X[i])

            # Mostramos el proceso
            print(f"Entrada: {X[i]} | Suma ponderada = {weighted_sum}")

            # Si llega o pasa el umbral → se activa (1)
            if weighted_sum >= self.umbral:
                predictions.append(1)
            else:
                predictions.append(0)

        return predictions

# 4. Ejecutamos la neurona
mcp = McCullochPitts(W, umbral)
predictions = mcp.predict(X)

print("\n✅ Predicciones finales:", predictions)
print("🎯 Salida esperada:     ", list(Y))
```

```python
import numpy as np

X = np.array([[0], [1]])  # Una sola entrada
Y = np.array([1, 0])      # Lo contrario

W = np.array([-1])  # Peso negativo
umbral = 0          # Umbral en 0

class McCullochPitts:
    def __init__(self, weights, umbral):
        self.weights = weights
        self.umbral = umbral

    def predict(self, X):
        predicciones = []
        for i in range(X.shape[0]):
            suma = np.dot(self.weights, X[i])
            print(f"Entrada: {X[i][0]} | Suma = {suma}")
            predicciones.append(1 if suma >= self.umbral else 0)
        return predicciones

neurona = McCullochPitts(W, umbral)
resultado = neurona.predict(X)
print("\n✅ NOT →", resultado)
print("🎯 Esperado:", list(Y))
```
